# Task 1: Order Approval Logic

Write a function called approval_message(order_value) that:
- Returns "Needs manager approval" if the value is above 100,000
- Returns "Needs supervisor approval" if it is above 50,000
- Returns "Auto-approved" otherwise
- Test it with an order value of 72,500


```python
# Function to decide the approval level of an order
def approval_message(order_value):
    if order_value > 100000:
        return "Needs manager approval"
    elif order_value > 50000:
        return "Needs supervisor approval"
    else:
        return "Auto-approved"

# Test with 72,500
print(approval_message(72500))
```

# Task 3: Supplier Status Function

Create a function called check_supplier_status(supplier_name, on_time_rate) that returns the supplier's name with a status:
- "Reliable" if the on-time rate is 0.90 or higher
- "Monitor" if it is 0.80 or higher
- "At Risk" otherwise
- Test it with at least 3 suppliers


```python
def check_supplier_status(supplier_name, on_time_rate):
    if on_time_rate >= 0.90:
        status = "Reliable"
    elif on_time_rate >= 0.80:
        status = "Monitor"
    else:
        status = "At Risk"
    return supplier_name + ": " + status

print(check_supplier_status("SupplierA", 0.94))
print(check_supplier_status("SupplierB", 0.86))
print(check_supplier_status("SupplierE", 0.74))
```

# Task 4: Load the Supply Chain Dataset
Load the file into a list of rows using the csv module, then print:
- The column names
- The number of data rows
- The first 5 rows

```python
import csv

with open("supply_chain_dataset.csv", newline="") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

print("Column names:", list(rows[0].keys()))
print("Number of data rows:", len(rows))

print("First 5 rows:")
for row in rows[:5]:
    print(row)
```

# Task 5: Explore the Dataset

Using loops and if statements, answer:
1. How many rows of data are there?
2. How many different supplier names are written in the file? Print them.
3. What is the highest total_cost?
4. What is the lowest total_cost? Does it make sense?
5. How many orders have the delivery_status "Delivered"?

```python
# Total number of rows
total_rows = 0
for row in rows:
    total_rows += 1
print("Number of rows:", total_rows)

# Different supplier names as written
supplier_names = []
for row in rows:
    if row["supplier_name"] not in supplier_names:
        supplier_names.append(row["supplier_name"])
print("Different supplier names:", len(supplier_names))
print(supplier_names)

# Highest and lowest total_cost (skip empty values)
highest = None
lowest = None
for row in rows:
    if row["total_cost"] != "":
        cost = float(row["total_cost"])
        if highest is None or cost > highest:
            highest = cost
        if lowest is None or cost < lowest:
            lowest = cost
print("Highest total_cost:", highest)
print("Lowest total_cost:", lowest)

# 5. Delivered orders
delivered = 0
for row in rows:
    if row["delivery_status"] == "Delivered":
        delivered += 1
print("Delivered orders:", delivered)
```

Challenge 1: Approval Based on Order Value and On-Time Rate

Write a function that takes an order value and a supplier's on-time rate and returns an approval level based on both. Reject orders above 100,000 from suppliers below 0.80.

```python
def smart_approval(order_value, on_time_rate):
    if order_value > 100000 and on_time_rate < 0.80:
        return "Rejected"
    elif order_value > 100000:
        return "Needs manager approval"
    elif order_value > 50000:
        return "Needs supervisor approval"
    else:
        return "Auto-approved"

print(smart_approval(150000, 0.70))
print(smart_approval(150000, 0.95))
print(smart_approval(60000, 0.70))
print(smart_approval(8000, 0.90))
```

Challenge 2: Find all SupplierA orders, however the name is written

Clean each name with lower() and replace(" ", ""), then count the orders and add up their total cost.

```python
supplier_a_count = 0
supplier_a_total = 0

for row in rows:
    clean_name = row["supplier_name"].lower().replace(" ", "")
    if clean_name == "suppliera":
        supplier_a_count += 1
        if row["total_cost"] != "":
            supplier_a_total += float(row["total_cost"])

print("SupplierA orders:", supplier_a_count)
print("SupplierA total cost:", round(supplier_a_total, 2))
```

# Challenge 3: Find Empty Values

Loop through the dataset and find which columns have gaps, and how many.


```python
empty_counts = {}

for column in rows[0].keys():
    empty_counts[column] = 0

for row in rows:
    for column in row:
        if row[column] == "":
            empty_counts[column] += 1

for column in empty_counts:
    if empty_counts[column] > 0:
        print(column, ":", empty_counts[column], "empty values")
```
